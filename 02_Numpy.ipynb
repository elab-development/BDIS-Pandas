# 2. NumPy

NumPy je biblioteka otvorenog koda za Python izuzetno pogodna za probleme linearne algebre,
brze i efikasne operacije nad matricama ili velikim nizovima, a koristi se kao temelj za mnoge
biblioteke u PyData ekosistemu - zbiru biblioteka i okvira, primarno u Python-u, koji za fokus ima
nauku o podacima.

Više o biblioteci se može saznati na zvaničnom sajtu koji sadrži dokumentaciju NumPy biblioteke:
https://numpy.org/doc/stable/

Počnimo sa komandom za uvoz NumPy biblioteke koja nam omogućava da njene funkcionalnosti koristimo
u daljem kodu.

```python
import numpy as np
```

Samu biblioteku smo "imenovali" standardnom skraćenicom `np`, koja će nam pozivanje biblioteke olakšati u daljem kodu.

## Sekcija 1 - NumPy funkcionalnosti, NumPy nizovi

Jedan od gradivnih elemenata biblioteke su nizovi i matrice. Fokusirajmo se u prvom delu na nizove
i definišimo jedan niz tako što ćemo običnu Python listu konvertovati u NumPy niz.

```python
lista = [1, 2, 3]
niz_np = np.array(lista)
niz_np
```

Proverimo tip `niz_np` promenljive:

```python
type(niz_np)
```

Po tipu `niz_np` promenljive zaista zaključujemo da smo kreirali jedan NumPy niz.
Postoje brojne ugrađene funkcije za kreiranje niza pa ih možemo isprobati.

Funkcija `arange` vraća jednako raspodeljene vrednosti u datom intervalu, sa mogućnošću definisanja
koraka. Gornja granica se ne uzima kao deo niza.

```python
niz2 = np.arange(0, 10, 2)
niz2
```

Možemo kreirati niz sačinjen od nula ili jedinica uz pomoć funkcija `zeros` i `ones`:

```python
np.zeros(5)
```

```python
np.ones(5)
```

Možemo iskoristiti i `linspace` funkciju koja vraća jednako raspoređene brojeve u specificiranom
intervalu - dakle ne zadajemo korak kao treći argument, već sam broj elemenata koje želimo da uključimo.

```python
np.linspace(0, 10, 10)
```

Možemo kreirati niz pomoću funkcije `rand` koja pseudonasumično generiše brojeve koji su zapravo
slučajan uzorak uniformne raspodele u intervalu [0, 1).

```python
np.random.rand(5)
```

Funkcija `randn`, za razliku od `rand`, vraća uzorak iz normalne raspodele.

```python
np.random.randn(30)
```

Postoji i funkcija koja nam vraća slučajne celobrojne vrednosti:

```python
np.random.randint(1, 100, 10)
```

## Sekcija 2 - NumPy funkcionalnosti, korisne funkcije nad nizovima

Pokažimo neke od korisnih funkcija nad nizovima. Počnimo od funkcija `max`, `min`, `argmax`, `argmin`.
Funkcija `max` nalazi maksimalan element niza, funkcija `min` minimum, a funkcije `argmax` i `argmin`
indeks maksimalnog i minimalnog elementa respektivno.

```python
niz3 = np.array([10, 456, 6467, 2342356, 22, 13, 6, 3, 1])
```

```python
print(np.max(niz3), np.min(niz3), np.argmax(niz3), np.argmin(niz3))
```

`shape` je atribut koji NumPy nizovi poseduju. Vraća oblik odnosno dimenzije niza.
Kod jednodimenzionalnog niza praktično dobijamo njegovu dužinu, a kasnije je naročito koristan za matrice.

```python
niz3.shape
```

Isti broj elemenata možemo dobiti i funkcijom `len`.

```python
len(niz3)
```

Tip podataka u nizu možemo ispitati preko atributa `dtype`.

```python
niz3.dtype
```

Elementi niza mogu biti uklonjeni korišćenjem funkcije `delete`. Kao parametre prosleđujemo niz i
poziciju indeksa elementa koji želimo da izbacimo. Rezultat moramo upisati nazad u niz.

```python
print(niz3)
niz3 = np.delete(niz3, [-1])
print(niz3)
```

Ako želimo da vratimo poslednji element, možemo koristiti funkciju `append`.

```python
niz3 = np.append(niz3, 1)
niz3
```

Funkcija `insert` ubacuje element na indeks koji prosledimo kao drugi parametar. Prvi parametar je niz,
a treći vrednost koju ubacujemo.

```python
niz3 = np.insert(niz3, 4, 222)
print(niz3)
```

Potrebno je voditi računa o dužini niza. Sledeći primer namerno pokušava da ubaci element na indeks
koji je izvan granica niza i zato generiše `IndexError`.

```python
niz3 = np.insert(niz3, 12, 800)
```

## Sekcija 3 - NumPy funkcionalnosti, indeksiranje niza

NumPy niz je fleksibilna struktura i omogućava jednostavno pristupanje pojedinačnim elementima,
podnizovima i opsezima.

```python
niz3[0]
```

```python
niz3[-1]
```

Možemo selektovati neki opseg niza, na primer od prvog do petog ili od drugog do petog elementa.

```python
niz3[0:5]
```

```python
niz3[1:5]
```

Jedna od mogućnosti koju NumPy niz ima je **broadcasting** - dodeljivanje iste vrednosti elementima
u nekom opsegu.

```python
niz3[0:3] = 100
niz3
```

Važno je razumeti šta se dešava kada napravimo podniz i promenimo mu vrednosti.
NumPy slajs je pogled na iste podatke, a ne nezavisna kopija.

```python
niz4 = np.arange(0, 11)
niz4
```

```python
niz5 = niz4[0:6]
niz5
```

```python
niz5[:] = 123
niz5
```

```python
niz4
```

Vidimo da se menja i originalni niz - NumPy nizovi su promenljive (mutable) strukture podataka.
Podaci nisu kopirani, već podniz deli iste memorijske lokacije sa originalnim nizom.

Ako želimo nezavisnu kopiju, koristimo funkciju `copy`.

```python
niz5_copy = niz5.copy()
niz5_copy[:] = 1
print(niz5_copy)
print(niz5)
```

Još jedan koristan način selekcije je selekcija po uslovu. Na primer, elementi veći od 5:

```python
niz4 = np.arange(0, 11)  # resetujemo niz
niz4[niz4 > 5]
```

Možemo proslediti i više indeksa koje želimo da selektujemo.

```python
niz3[[0, 2, 4, 5]]
```

## Sekcija 4 - NumPy funkcionalnosti, operacije nad nizom

Nad nizovima se mogu vršiti aritmetičke operacije poput sabiranja, oduzimanja, množenja i deljenja.

```python
niz6 = np.arange(0, 10)
niz7 = np.arange(0, 10)
```

```python
niz6 + niz7
```

```python
niz6 - niz7
```

```python
niz6 * niz7
```

```python
niz6 / niz7
```

Prilikom deljenja sa nulom NumPy daje upozorenje i `nan`, ali ne prekida izvršavanje greškom.

Postoje i dodatne matematičke operacije poput kvadriranja, kvadratnog korena i slično.

```python
niz6**2
```

```python
np.sqrt(niz6)
```

```python
np.exp(niz6)
```

```python
np.sin(niz6)
```

```python
np.log(niz6)
```

Naravno, možemo i množiti vektor (NumPy niz) skalarom.

```python
niz6 * 10
```

**Bonus pitanje:** Šta bi bio rezultat da običnu Python listu pomnožimo skalarom?

Možemo i deliti skalarom, sabirati i oduzimati skalar od vektora - operacija se izvršava nad svakim
elementom pojedinačno.

```python
print(niz6 / 10)
print(niz6 - 10)
print(niz6 + 100)
```

NumPy funkcijom `dot` omogućeno je vektorsko množenje dva niza.

```python
print(niz6.dot(niz7))
# Možemo ga računati i na sledeći način:
print(np.sum(niz6 * niz7))
```

## Sekcija 5 - NumPy funkcionalnosti, NumPy matrice

NumPy sadrži implementaciju matrica koje su zapravo višedimenzionalni nizovi tipa `numpy.ndarray`.
Biblioteka omogućava brojne operacije nad matricama: množenje i sabiranje matrica i skalara,
inverziju, transponovanje itd.

Definišimo jednu 2x2 matricu:

```python
matrica = np.array([[44, 29], [18, 87]])
print(matrica)
```

Ispišimo dimenzije matrice preko atributa `shape`:

```python
matrica.shape
```

Pokažimo razliku između element-wise i dot (matričnog) množenja:

```python
print(matrica * matrica)
print("***********")
print(matrica.dot(matrica))
```

Možemo množiti matricu skalarom ili vektorom po elementima:

```python
vektor = np.array([2, 2])
print(matrica * vektor)
print(matrica * 10)
```

Pazeći na dimenzije, možemo računati i proizvod matrice i vektora:

```python
matrica.dot(vektor)
```

Možemo sabirati i oduzimati matrice po elementima:

```python
print(matrica + matrica)
print(matrica - matrica)
```

Možemo dodavati redove i kolone matrici, pazeći na dimenzije.

```python
matrica = np.vstack((matrica, np.array([2, 2])))
print(matrica)
```

```python
matrica = np.hstack((matrica, np.array([[3], [3], [3]])))
print(matrica)
```

Funkcijom `vstack` možemo dodati redove matrici, dok funkcijom `hstack` dodajemo kolone.

Matrice možemo generisati i pomoću funkcija `zeros`, `ones`, `random` i slično.

```python
print(np.ones((5, 5)))
print(np.zeros((5, 5)))
print(np.random.random((5, 5)))
```

Slajseri rade veoma slično kao i kod niza, uz činjenicu da matrica ima još jednu dimenziju.

```python
matrica[:, :]  # ili samo matrica[:]
```

Izbor trećeg reda:

```python
matrica[2, :]
```

Izbor treće kolone:

```python
matrica[:, 2]
```

Element drugog reda i druge kolone:

```python
matrica[1, 1]
```

Podmatrica koju čine prvi i drugi red i prva i druga kolona:

```python
matrica[0:2, 0:2]
```

Izbor elemenata po uslovu, na primer vrednosti veće od 10:

```python
matrica[matrica > 10]
```
