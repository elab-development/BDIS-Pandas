## Sekcija 1 - Matplotlib

Počnimo od komande za import Matplotlib biblioteke:

```python
import matplotlib.pyplot as plt
```

```python
import warnings
warnings.filterwarnings('ignore')
```

Prođimo kroz jednostavan primer. Želimo da iscrtamo grafik za dva NumPy niza.
Jedan predstavlja koordinate x, a drugi koordinate y ose.

```python
import numpy as np
x = np.linspace(0, 5, 11)
y = x**2
```

Definišimo sada grafik za naše nizove:

```python
plt.plot(x, y, 'r')  # parametar r označava crvenu liniju
plt.xlabel('Ovo je x osa')
plt.ylabel('Ovo je y osa')
plt.title('Ovo je naslov našeg grafika!')
plt.show()
```

Možemo kreirati i više grafika tako što ćemo koristiti funkciju `subplot`.

Potpis funkcije:

`subplot(broj_redova_na_grafiku, broj_kolona_na_grafiku, broj_grafika)`

```python
plt.subplot(1, 2, 1)
plt.plot(x, y, 'r--')

plt.subplot(1, 2, 2)
plt.plot(x, y, 'g*-')
```

Matplotlib podržava metod rada koji podseća na objektno orijentisano programiranje.
Možemo instancirati objekat tipa `figure` i zatim pozivati metode ili atribute tog objekta.

```python
fig = plt.figure()
axes = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes.plot(x, y, 'b')
axes.set_xlabel('Ovo je x osa')
axes.set_ylabel('Ovo je y osa')
axes.set_title('Ovo je naslov grafika')
```

`Figure` predstavlja "platno" na kom iscrtavamo grafik. Funkcija `add_axes` dodaje ose na grafik.
Parametri u listi predstavljaju položaj od leve ivice i dna, kao i širinu i visinu osa u odnosu na figuru.

`subplots` funkcija omogućava automatizovanije kreiranje osa i lako kreiranje više grafika odjednom.

```python
fig, axes = plt.subplots()
axes.plot(x, y, 'r')
axes.set_xlabel('Ovo je x osa.')
axes.set_ylabel('Ovo je y osa.')
axes.set_title('Naslov našeg grafika.')
```

Ovako možemo kreirati više grafika i iterirati kroz njih:

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
for ax in axes:
    ax.plot(x, y, 'b')
    ax.set_xlabel('Ovo je x osa.')
    ax.set_ylabel('Ovo je y osa.')
    ax.set_title('Naslov našeg grafika.')
```

Ako se sadržaj preklapa, možemo pozvati `tight_layout()` da bismo sprečili da sadržaj jednog grafika
prekriva drugi.

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
for ax in axes:
    ax.plot(x, y, 'b')
    ax.set_xlabel('Ovo je x osa.')
    ax.set_ylabel('Ovo je y osa.')
    ax.set_title('Naslov našeg grafika.')

plt.tight_layout()
```

Parametrom `figsize` možemo podešavati širinu i visinu figure:

```python
fig = plt.figure(figsize=(8, 4))
axes = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes.plot(x, y, 'b')
axes.set_xlabel('Ovo je x osa')
axes.set_ylabel('Ovo je y osa')
axes.set_title('Ovo je naslov grafika')
```

Matplotlib može da čuva grafike u velikom broju formata - PNG, JPG, EPS, SVG, PGF i PDF.

```python
fig.savefig("slikaGrafika.png")
```

Pored naslova i oznaka osa, grafik može imati i legendu. Pozicija legende se može menjati parametrom
`loc`; često se koristi `loc=0` kako bi Matplotlib sam izabrao odgovarajuću poziciju.

```python
fig = plt.figure()
axes = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes.set_xlabel('Ovo je x osa')
axes.set_ylabel('Ovo je y osa')
axes.set_title('Ovo je naslov grafika')
axes.plot(x, x**2, label="y=x^2")
axes.plot(x, x**3, label="y=x^3")
axes.legend()
plt.show()
```

Matplotlib nudi opcije za podešavanje tipova i boja linija. Na primer, `'b'` je plava, `'g'` zelena,
dok stil `'b.-'` predstavlja plavu liniju sa tačkicama.

```python
fig, ax = plt.subplots()
ax.plot(x, x**2, 'b.-')
ax.plot(x, x**3, 'g--')
```

Boja se može podesiti i eksplicitnim dodeljivanjem vrednosti parametru `color`, koji prima naziv boje
ili heksadecimalnu vrednost. Parametar `alpha` reguliše transparentnost.

```python
fig, ax = plt.subplots()
ax.plot(x, x+1, color="blue", alpha=0.1)
ax.plot(x, x+2, color="#8B008B")
ax.plot(x, x+3, color="#FF8C00")
```

Matplotlib pruža opcije za podešavanje granica vrednosti na y osi pomoću `set_ylim` i na x osi pomoću
`set_xlim`.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(x, x**2, x, x**3)
axes[0].set_title("Podrazumevani opseg osa.")

axes[1].plot(x, x**2, x, x**3)
axes[1].set_ylim([0, 60])
axes[1].set_xlim([2, 5])
axes[1].set_title("Zadati opseg.")
```

Postoje i druge vrste grafika poput scatter plot-a, histograma ili boxplot-a.

```python
plt.scatter(x, y)
```

```python
from random import sample
data = sample(range(1, 1000), 100)
plt.hist(data)
```

```python
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

# rectangular box plot
plt.boxplot(data, vert=True, patch_artist=True);
```

## Seaborn

Seaborn je još jedna biblioteka pogodna za vizualizaciju podataka sa kojom možemo iscrtavati
kompleksnije vrste grafikona. Počećemo od grafikona koji dobro vizualizuju raspodele promenljivih.

Za primere koristimo `tips` dataset koji Seaborn biblioteka sadrži.

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

Funkcija `displot` služi za vizualizaciju raspodele jedne promenljive:

```python
sns.displot(tips['total_bill'])
```

```python
sns.displot(tips['total_bill'], kde=False, bins=50)
```

Broj `bin`-ova utiče na izgled prikaza raspodele. Za diskretne numeričke podatke možemo koristiti
parametar `discrete=True`.

```python
sns.displot(tips['size'], discrete=True)
```

Raspodelu možemo prikazati i po kategorijama, na primer po polu:

```python
sns.displot(tips, x='total_bill', hue='sex', multiple='stack')
```

`jointplot` daje vizualizaciju po dve promenljive - jedna je na x, druga na y osi.

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='scatter')
```

Možemo uočavati trendove i potencijalne korelacije. `pairplot` kreira grafik za svaki par kolona i može
pomoći u uočavanju korelisanosti u podacima.

```python
sns.pairplot(tips)
```

Podatke možemo dodatno podeliti po kategoriji, u ovom slučaju po polu.

```python
sns.pairplot(tips, hue='sex', palette='coolwarm')
```

KDE je skraćenica za Kernel Density Estimation - metod za procenu raspodele numeričke vrednosti.
KDE grafik aproksimira raspodelu na osnovu svih opservacija.

```python
sns.kdeplot(tips['total_bill'])
```

Parametrom `bw_adjust` možemo menjati nivo zaglađivanja (smoothing):

```python
sns.kdeplot(data=tips, x="total_bill", bw_adjust=.2)
```

```python
sns.kdeplot(data=tips, x="total_bill", bw_adjust=5, cut=0)
```

KDE prikaz možemo razdvojiti po kategoriji, na primer po vremenu obroka:

```python
sns.kdeplot(data=tips, x="total_bill", hue="time", multiple="stack")
```

Prosleđivanjem `cumulative=True` dobijamo kumulativnu funkciju raspodele.

```python
import scipy
```

```python
sns.kdeplot(tips["total_bill"], cumulative=True)
```

Podatke možemo vizualizovati po kategorijama. `barplot` može prikazati srednju vrednost neke numeričke
promenljive po kategoriji.

```python
sns.barplot(x='sex', y='total_bill', data=tips)
```

`barplot`-u možemo proslediti i drugu agregacionu funkciju preko parametra `estimator`, na primer
standardnu devijaciju.

```python
import numpy as np
sns.barplot(x='sex', y='total_bill', data=tips, estimator=np.std)
```

`countplot` je sličan barplot-u, ali grupiše podatke i računa broj opservacija po kategoriji.

```python
sns.countplot(x='sex', data=tips)
```

Boxplot možemo koristiti da pogledamo razlike u raspodeli numeričke promenljive po kategoriji i
identifikujemo potencijalne autlajere.

```python
sns.boxplot(x="day", y="total_bill", data=tips, palette="rainbow")
```

Možemo dodatno podeliti podatke i po drugoj kategoriji, na primer da li je osoba pušač ili ne:

```python
sns.boxplot(x="day", y="total_bill", data=tips, hue="smoker", palette="coolwarm")
```

Grafik sličan boxplot-u je `violinplot`, gde se koristi KDE pristup za procenu raspodele.

```python
sns.violinplot(x="day", y="total_bill", data=tips, palette="rainbow")
```

```python
sns.violinplot(x="day", y="total_bill", data=tips, hue="sex", palette="Set1")
```

```python
sns.violinplot(x="day", y="total_bill", data=tips, hue="sex", split=True, palette="Set1")
```

Veoma informativna vizualizacija je `heatmap`. Ona može da prikaže korelacionu matricu numeričkih
atributa.

```python
tips_num = tips.select_dtypes(exclude=["category"])
```
