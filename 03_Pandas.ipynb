## Sekcija 2 Pandas - selekcija i "slajsovanje" DataFrame strukture

Određene kolone iz DataFrame-a možemo selektovati pomoću njihovog naziva:

```python
df['BasePay']
```

Možemo uzeti i listu kolona:

```python
df[['BasePay', 'OvertimePay', 'OtherPay']]
```

Određenu kolonu možemo izbaciti funkcijom `drop`. Rezultat moramo upisati nazad u promenljivu `df`
ili specificirati `inplace=True`.

```python
df = df.drop('TotalPay', axis=1)  # axis=1 označava da biramo kolonu
# alternativa: df.drop('TotalPay', axis=1, inplace=True)
```

```python
df.head(3)
```

Želimo da kreiramo novu kolonu. Prosleđujemo njen naziv i sabiramo `BasePay`, `OvertimePay` i `OtherPay`.

```python
df['TotalPayNew'] = df['BasePay'] + df['OvertimePay'] + df['OtherPay']
df.head(3)
```

Redove možemo birati po logičkom uslovu. Na primer, instance čiji je `BasePay` veći od 200000:

```python
df[df['BasePay'] > 200000]
```

Možemo zadavati više logičkih uslova:

```python
df[(df['BasePay'] > 200000) & (df['JobTitle'].str.contains('Assistant'))]
```

Možemo zadati logički uslov i izabrati samo jedan deo kolona:

```python
df[(df['BasePay'] > 200000) & (df['JobTitle'].str.contains('Assistant'))][['EmployeeName', 'JobTitle']]
```

Još jedan značajan način za selekciju određenih redova ili kolona su `loc` i `iloc`.

- `loc` koristi nazive kolona i/ili oznake redova.
- `iloc` koristi poziciju odnosno redosled.

```python
df.loc[0:20, ['EmployeeName', 'JobTitle']]
```

Ukoliko ne znamo nazive kolona, ali znamo njihove pozicije, koristimo `iloc`.

```python
df.iloc[[0, 57, 534, 666], [1, 2]]
```

## Sekcija 3 Pandas - nedostajuće vrednosti

Ispitajmo naš DataFrame još jednom:

```python
df.head()
```

`NaN` (Not a Number) predstavlja nedostajuće ili nepoznate vrednosti. Funkcijom `isna` možemo da
proverimo strukturu nedostajućih vrednosti u celom DataFrame-u.

```python
df.isna().sum()
```

Nedostajuće numeričke podatke možemo zameniti srednjom vrednošću odgovarajuće kolone.

```python
df['BasePay'].fillna(value=df['BasePay'].mean(), inplace=True)
df['TotalPayNew'].fillna(value=df['TotalPayNew'].mean(), inplace=True)
df['Benefits'].fillna(value=df['Benefits'].mean(), inplace=True)

df['OvertimePay'].fillna(value=df['OvertimePay'].mean(), inplace=True)
df['OtherPay'].fillna(value=df['OtherPay'].mean(), inplace=True)
```

Za kategoričke podatke `Notes` i `Status` ima previše nedostajućih vrednosti, pa u ovom primeru
izbacujemo te kolone.

```python
df.drop('Notes', axis=1, inplace=True)
df.drop('Status', axis=1, inplace=True)
```

```python
print(f'Broj nedostajućih vrednosti posle zamene: {df.isna().sum().sum()}.')
```

## Sekcija 5 Pandas - kombinovanje DataFrame-a

Postoje tri glavna načina za kombinovanje DataFrame-a: **merging**, **joining** i **concatenating**.
Daćemo primere za svaki.

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B': ['B0', 'B1', 'B2', 'B3'],
    'C': ['C0', 'C1', 'C2', 'C3'],
    'D': ['D0', 'D1', 'D2', 'D3']
}, index=[0, 1, 2, 3])

df2 = pd.DataFrame({
    'A': ['A4', 'A5', 'A6', 'A7'],
    'B': ['B4', 'B5', 'B6', 'B7'],
    'C': ['C4', 'C5', 'C6', 'C7'],
    'D': ['D4', 'D5', 'D6', 'D7']
}, index=[4, 5, 6, 7])

df3 = pd.DataFrame({
    'A': ['A8', 'A9', 'A10', 'A11'],
    'B': ['B8', 'B9', 'B10', 'B11'],
    'C': ['C8', 'C9', 'C10', 'C11'],
    'D': ['D8', 'D9', 'D10', 'D11']
}, index=[8, 9, 10, 11])
```

```python
df1
```

```python
df2
```

```python
df3
```

**Concatenation** praktično lepi DataFrame-ove jedan za drugi. Potrebno je voditi računa o dimenzijama
i osi po kojoj vršimo spajanje.

```python
pd.concat([df1, df2, df3])
```

```python
pd.concat([df1, df2, df3], axis=1)
```

`merge` radi sličnom logikom kao spajanje dve SQL tabele. Definišimo dva DataFrame-a sa zajedničkim
ključem.

```python
left = pd.DataFrame({
    'key': ['K0', 'K1', 'K2', 'K3'],
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B': ['B0', 'B1', 'B2', 'B3']
})

right = pd.DataFrame({
    'key': ['K0', 'K1', 'K2', 'K3'],
    'C': ['C0', 'C1', 'C2', 'C3'],
    'D': ['D0', 'D1', 'D2', 'D3']
})
```

```python
left
```

```python
right
```

```python
pd.merge(left, right, how='inner', on='key')
```

**Joining** je pogodan način za spajanje kolona dve tabele koje su potencijalno drugačije indeksirane
u jedan DataFrame.

```python
left = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
}, index=['K0', 'K1', 'K2'])

right = pd.DataFrame({
    'C': ['C0', 'C2', 'C3'],
    'D': ['D0', 'D2', 'D3']
}, index=['K0', 'K2', 'K3'])
```

```python
left.join(right)
```

```python
left.join(right, how='outer')
```
